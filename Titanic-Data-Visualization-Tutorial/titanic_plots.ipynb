```python
# Import Libraries
import pandas as pd 
import seaborn as sb
import matplotlib.pyplot as plt 
```

```python
# Load the built-in Titanic dataset 
df=sb.load_dataset('titanic')

# Inspect the first 5 rows
print(df.head())
```

```python
# Count Plot (categorical distribution)
sb.countplot(data=df, x='class', hue='survived') #Shows the exact count of passengers in each class, split by survival.
```

```python
# Create a barplot showing survival rate by class
sb.barplot(data=df, x='class', y='survived', hue='sex')
plt.title("Titanic Survival Rate by Class and Sex")
plt.show()
```

```python
# Create a barplot showing survival rate by class
sb.boxplot(data=df, x='class', y='age', hue='sex')
plt.show()
```
